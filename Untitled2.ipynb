# 8. Implementación del perceptrón clásico

El perceptrón es uno de los modelos más básicos de una red neuronal artificial.

Su funcionamiento consiste en realizar una combinación lineal de las entradas mediante pesos y un sesgo o `bias`.

La expresión utilizada es:

z = X · W + b

Posteriormente se aplica una función de activación escalón para obtener una clasificación binaria.

En esta implementación los pesos se inicializan en cero y se actualizan cuando existe un error de clasificación.

```python
def funcion_escalon(z):
    return np.where(z >= 0, 1, 0)


def entrenar_perceptron(X, y, learning_rate=0.1, epochs=100):

    pesos = np.zeros(X.shape[1])
    bias = 0.0

    historial_errores = []

    for epoch in range(epochs):

        errores = 0

        for xi, yi in zip(X, y):

            z = np.dot(xi, pesos) + bias

            prediccion = funcion_escalon(z)

            error = yi - prediccion

            pesos += learning_rate * error * xi
            bias += learning_rate * error

            if error != 0:
                errores += 1

        historial_errores.append(errores)

        if epoch % 10 == 0:
            print(
                "Época:", epoch,
                "| Errores:", errores
            )

        if errores == 0:
            print(
                "El perceptrón convergió en la época:",
                epoch
            )
            break

    return pesos, bias, historial_errores
```

## Entrenamiento del perceptrón

A continuación se entrena el perceptrón utilizando los datos de entrenamiento.

El algoritmo realiza sucesivas actualizaciones de los pesos y del bias. En cada época se cuentan los errores de clasificación.

Cuando el número de errores llega a cero, el algoritmo puede detenerse porque alcanzó una solución para los datos utilizados.

```python
pesos_perceptron, bias_perceptron, errores_perceptron = entrenar_perceptron(
    X_train,
    y_train,
    learning_rate=0.1,
    epochs=100
)

print("\nPesos finales:")
print(pesos_perceptron)

print("\nBias final:")
print(bias_perceptron)
```

# 11. Función de activación Sigmoide

La función Sigmoide transforma un valor real en un valor comprendido entre 0 y 1.

Su expresión matemática es:

σ(z) = 1 / (1 + e^(-z))

Debido a que produce valores entre 0 y 1, es especialmente útil para problemas de clasificación binaria, donde puede interpretarse la salida como una probabilidad de pertenecer a una determinada clase.

```python
def sigmoid(z):
    """
    Función de activación Sigmoid.
    """
    return 1 / (1 + np.exp(-z))
```

# 12. Función de pérdida: Binary Cross-Entropy

Para evaluar qué tan diferentes son las predicciones del modelo respecto a los valores reales se utiliza Binary Cross-Entropy.

Esta función es apropiada para problemas de clasificación binaria.

El objetivo del entrenamiento será minimizar este error mediante la actualización progresiva de los pesos y del bias.

```python
def calcular_error(y_real, y_predicha):

    epsilon = 1e-8

    return -np.mean(
        y_real * np.log(y_predicha + epsilon)
        +
        (1 - y_real) * np.log(
            1 - y_predicha + epsilon
        )
    )
```

# 13. Red neuronal de una capa

A diferencia del perceptrón clásico, en esta sección se implementa una neurona utilizando una función de activación Sigmoide y una función de pérdida.

La arquitectura recibe tres características de entrada y produce una única salida.

La salida representa la probabilidad estimada de que la fruta pertenezca a la clase BANANO.

La estructura es:

3 características de entrada → neurona → Sigmoide → probabilidad

```python
np.random.seed(42)

W = np.random.randn(3, 1) * 0.01
b = 0.0

print("Pesos iniciales:")
print(W)

print("Bias inicial:")
print(b)
```

# 14. Forward Propagation

La propagación hacia adelante consiste en llevar los datos desde la entrada hasta la salida de la red.

Primero se calcula:

Z = XW + b

Después se aplica la función Sigmoide:

A = Sigmoid(Z)

El resultado `A` representa la predicción de la red.

```python
Z = X_train @ W + b

A = sigmoid(Z)

print("Primeras predicciones:")
print(A[:5])

error_inicial = calcular_error(
    y_train.reshape(-1, 1),
    A
)

print("Error inicial:")
print(error_inicial)
```

# 15. Backpropagation: una iteración

Backpropagation es el procedimiento utilizado para calcular cómo deben modificarse los parámetros de la red para reducir el error.

En este proceso se calcula el gradiente del error respecto a los pesos y al bias.

Primero se calcula:

dZ = A - y

Después se obtiene el gradiente de los pesos:

dW = Xᵀ · dZ / m

Y el gradiente del bias:

db = promedio(dZ)

Finalmente, los parámetros se actualizan utilizando el descenso del gradiente:

W = W - α dW

b = b - α db

donde α representa la tasa de aprendizaje.

```python
y_train_col = y_train.reshape(-1, 1)

dZ = A - y_train_col

dW = X_train.T @ dZ / len(X_train)

db = np.mean(dZ)

learning_rate = 0.1

W = W - learning_rate * dW
b = b - learning_rate * db

print("Pesos después de una actualización:")
print(W)

print("Bias después de una actualización:")
print(b)
```

# 16. Entrenamiento completo de la red de una capa

En la sección anterior se realizó una sola actualización de los parámetros.

Ahora se repite el proceso durante 2000 épocas.

Cada época contiene cuatro etapas principales:

1. Forward Propagation.
2. Cálculo de la función de pérdida.
3. Backpropagation.
4. Actualización de pesos y bias.

El error obtenido en cada época se almacena en `historial_loss` para posteriormente analizar gráficamente el aprendizaje de la red.

```python
np.random.seed(42)

W = np.random.randn(3, 1) * 0.01
b = 0.0

learning_rate = 0.1
epochs = 2000

historial_loss = []

for epoch in range(epochs):

    Z = X_train @ W + b

    A = sigmoid(Z)

    error = calcular_error(
        y_train_col,
        A
    )

    historial_loss.append(error)

    dZ = A - y_train_col

    dW = X_train.T @ dZ / len(X_train)

    db = np.mean(dZ)

    W = W - learning_rate * dW

    b = b - learning_rate * db
```

# 19. Función de activación ReLU

La función ReLU significa Rectified Linear Unit.

Su funcionamiento es:

ReLU(z) = max(0,z)

Por lo tanto:

- Si z es negativo, la salida es 0.
- Si z es positivo, la salida conserva su valor.

ReLU es ampliamente utilizada en capas ocultas de redes neuronales porque permite introducir no linealidad en el modelo.

```python
def relu(z):
    """
    Función de activación ReLU.
    """
    return np.maximum(0, z)


def derivada_relu(z):
    """
    Derivada de ReLU.
    """
    return (z > 0).astype(float)

```

# 20. Red neuronal multicapa

En esta sección se implementa manualmente una red neuronal multicapa.

La arquitectura utilizada es:

3 entradas → 4 neuronas ocultas → 1 neurona de salida

La capa oculta utiliza ReLU y la capa de salida utiliza Sigmoide.

Esta arquitectura permite representar relaciones más complejas que un modelo formado únicamente por una neurona.

```python
print("\n" + "=" * 60)
print("RED NEURONAL MULTICAPA MANUAL")
print("=" * 60)


np.random.seed(42)

W1 = np.random.randn(
    3,
    4
) * np.sqrt(2 / 3)

b1 = np.zeros(
    (1, 4)
)

W2 = np.random.randn(
    4,
    1
) * np.sqrt(2 / 4)

b2 = np.zeros(
    (1, 1)
)

print("\nDimensión W1:", W1.shape)
print("Dimensión b1:", b1.shape)

print("Dimensión W2:", W2.shape)
print("Dimensión b2:", b2.shape)
```

# 21. Forward Propagation de la red multicapa

La información pasa primero por la capa oculta:

Z1 = XW1 + b1

Luego se aplica ReLU:

A1 = ReLU(Z1)

Posteriormente se calcula la salida:

Z2 = A1W2 + b2

Finalmente se aplica Sigmoide:

A2 = Sigmoid(Z2)

De esta manera se obtiene la probabilidad correspondiente a la clasificación.

```python
Z1 = X_train @ W1 + b1

A1 = relu(Z1)

Z2 = A1 @ W2 + b2

A2 = sigmoid(Z2)

print("Salida de la capa oculta:")
print(A1[:5])

print("Primeras probabilidades:")
print(A2[:5])
```

# 22. Backpropagation de la red multicapa

En una red multicapa, el error debe propagarse desde la salida hacia las capas anteriores.

Primero se calcula el gradiente correspondiente a la capa de salida.

Posteriormente el error se propaga hacia la capa oculta mediante la multiplicación por los pesos de la segunda capa.

Finalmente se utiliza la derivada de ReLU para determinar cómo actualizar los parámetros de la capa oculta.

Este proceso permite calcular los gradientes de:

- W2
- b2
- W1
- b1

```python
m = len(X_train)

# Gradiente de la capa de salida

dZ2 = A2 - y_train_col

dW2 = A1.T @ dZ2 / m

db2 = np.mean(
    dZ2,
    axis=0,
    keepdims=True
)

# Propagación del error hacia la capa oculta

dA1 = dZ2 @ W2.T

dZ1 = dA1 * derivada_relu(Z1)

dW1 = X_train.T @ dZ1 / m

db1 = np.mean(
    dZ1,
    axis=0,
    keepdims=True
)

print("\nGradientes calculados correctamente.")
```

# 23. Entrenamiento de la red multicapa

La red se entrena durante 3000 épocas.

En cada época se realizan las siguientes operaciones:

1. Forward Propagation.
2. Cálculo del error.
3. Backpropagation.
4. Cálculo de gradientes.
5. Actualización de pesos y bias.

El error de cada época se almacena para poder analizar posteriormente la evolución del aprendizaje.

```python
np.random.seed(42)

W1 = np.random.randn(
    3,
    4
) * np.sqrt(2 / 3)

b1 = np.zeros(
    (1, 4)
)

W2 = np.random.randn(
    4,
    1
) * np.sqrt(2 / 4)

b2 = np.zeros(
    (1, 1)
)

learning_rate = 0.05
epochs = 3000

historial_loss_multicapa = []

for epoch in range(epochs):

    # ----------------------------------------
    # Forward
    # ----------------------------------------

    Z1 = X_train @ W1 + b1

    A1 = relu(Z1)

    Z2 = A1 @ W2 + b2

    A2 = sigmoid(Z2)

    # ----------------------------------------
    # Loss
    # ----------------------------------------

    error = calcular_error(
        y_train_col,
        A2
    )

    historial_loss_multicapa.append(
        error
    )

    # ----------------------------------------
    # Backpropagation
    # ----------------------------------------

    m = len(X_train)

    dZ2 = A2 - y_train_col

    dW2 = A1.T @ dZ2 / m

    db2 = np.mean(
        dZ2,
        axis=0,
        keepdims=True
    )

    dA1 = dZ2 @ W2.T

    dZ1 = dA1 * derivada_relu(Z1)

    dW1 = X_train.T @ dZ1 / m

    db1 = np.mean(
        dZ1,
        axis=0,
        keepdims=True
    )

    # ----------------------------------------
    # Actualización
    # ----------------------------------------

    W2 -= learning_rate * dW2

    b2 -= learning_rate * db2

    W1 -= learning_rate * dW1

    b1 -= learning_rate * db1

    if epoch % 500 == 0:

        print(
            "Época:",
            epoch,
            "- Error:",
            round(error, 6)
        )
```

## Función Softmax

Para la clasificación multiclase se utiliza una capa de salida con tres neuronas y función de activación Softmax.

Softmax transforma las salidas de la red en probabilidades que suman 1.

# 29. Normalización de las imágenes

Los valores originales de los píxeles de las imágenes están comprendidos entre 0 y 255.

Para facilitar el entrenamiento, estos valores se convierten a `float32` y se dividen entre 255.

De esta manera los valores quedan normalizados en un rango entre 0 y 1.

Para MNIST se utiliza una red neuronal completamente conectada.

# 33. Implementación de MNIST con PyTorch

Para comparar diferentes frameworks de Deep Learning, se implementa nuevamente un modelo para MNIST utilizando PyTorch.

Primero se convierten los datos de NumPy a tensores de PyTorch.

Posteriormente se construye un Dataset y un DataLoader para organizar los datos en lotes o batches durante el entrenamiento.

```python
x_train_torch = torch.tensor(
    x_train_mnist,
    dtype=torch.float32
)

y_train_torch = torch.tensor(
    y_train_mnist,
    dtype=torch.long
)

x_test_torch = torch.tensor(
    x_test_mnist,
    dtype=torch.float32
)

y_test_torch = torch.tensor(
    y_test_mnist,
    dtype=torch.long
)

print(
    "Tensor de entrenamiento:",
    x_train_torch.shape
)

print(
    "Tensor de prueba:",
    x_test_torch.shape
)
train_dataset = TensorDataset(
    x_train_torch,
    y_train_torch
)

test_dataset = TensorDataset(
    x_test_torch,
    y_test_torch
)

train_loader = DataLoader(
    train_dataset,
    batch_size=64,
    shuffle=True
)

test_loader = DataLoader(
    test_dataset,
    batch_size=64,
    shuffle=False
)

print(
    "\nDataLoaders creados correctamente."
)

```

# 34. Definición de la red neuronal en PyTorch

La red neuronal se define mediante una clase que hereda de `nn.Module`.

La arquitectura utilizada es:

784 entradas
↓
128 neuronas
↓
ReLU
↓
10 salidas

Las 784 entradas corresponden a los 28 × 28 píxeles de cada imagen.

La salida contiene 10 valores correspondientes a los diez posibles dígitos.

```python
class RedMNIST(nn.Module):

    def __init__(self):

        super().__init__()

        self.flatten = nn.Flatten()

        self.capa1 = nn.Linear(
            784,
            128
        )

        self.relu = nn.ReLU()

        self.salida = nn.Linear(
            128,
            10
        )


    def forward(self, x):

        x = self.flatten(x)

        x = self.capa1(x)

        x = self.relu(x)

        x = self.salida(x)

        return x
```

# 35. Entrenamiento y Backpropagation en PyTorch

Durante el entrenamiento se realizan cinco pasos principales:

1. Se limpian los gradientes acumulados.
2. Se realiza la propagación hacia adelante.
3. Se calcula la función de pérdida.
4. Se ejecuta Backpropagation mediante `error.backward()`.
5. Se actualizan los pesos mediante `optimizador.step()`.

A diferencia de las implementaciones manuales anteriores, PyTorch calcula automáticamente los gradientes utilizando su sistema de diferenciación automática.

```python
modelo_pytorch = RedMNIST()

print(modelo_pytorch)
```

```python
criterio = nn.CrossEntropyLoss()

optimizador = optim.Adam(
    modelo_pytorch.parameters(),
    lr=0.001
)

print(
    "\nModelo, criterio y optimizador listos."
)
```

```python
epochs = 3

historial_error_pytorch = []

for epoch in range(epochs):

    modelo_pytorch.train()

    error_total = 0

    for imagenes, etiquetas in train_loader:

        # ------------------------------------
        # 1. Limpiar gradientes
        # ------------------------------------

        optimizador.zero_grad()

        # ------------------------------------
        # 2. Forward
        # ------------------------------------

        salidas = modelo_pytorch(
            imagenes
        )

        # ------------------------------------
        # 3. Loss
        # ------------------------------------

        error = criterio(
            salidas,
            etiquetas
        )

        # ------------------------------------
        # 4. Backpropagation
        # ------------------------------------

        error.backward()

        # ------------------------------------
        # 5. Actualizar pesos
        # ------------------------------------

        optimizador.step()

        error_total += error.item()

    promedio_error = (
        error_total / len(train_loader)
    )

    historial_error_pytorch.append(
        promedio_error
    )

    print(
        "Época:",
        epoch + 1,
        "- Error:",
        round(promedio_error, 4)
    )
```

# 36. Evaluación del modelo PyTorch

Una vez finalizado el entrenamiento, el modelo se coloca en modo evaluación mediante `model.eval()`.

Posteriormente se desactivan los cálculos de gradientes con `torch.no_grad()`, ya que durante la evaluación no es necesario actualizar los parámetros.

Se comparan las predicciones obtenidas con las etiquetas reales y se calcula el Accuracy.

```python
modelo_pytorch.eval()

correctas = 0

total = 0

with torch.no_grad():

    for imagenes, etiquetas in test_loader:

        salidas = modelo_pytorch(
            imagenes
        )

        _, predicciones = torch.max(
            salidas,
            1
        )

        total += etiquetas.size(0)

        correctas += (
            predicciones == etiquetas
        ).sum().item()


accuracy_pytorch = correctas / total

print(
    "\nAccuracy en prueba con PyTorch:",
    accuracy_pytorch
)
```
